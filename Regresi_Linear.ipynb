#### Import Library

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score

#202431036_Fathur Rasyid Irimawa
```

#### Membaca Dataset hasil Scraping

```python
df = pd.read_csv("data_utang_negara_clean.csv")
print(f"Jumlah baris awal: {len(df)}")
print(df.head(3))

#202431036_Fathur Rasyid Irimawa
```

#### Membuang data kosong yang ada pada kolom x dan y

```python
df = df.dropna(subset=['Per_Capita', 'GDP_Percentage', 'External_Debt'])

#202431036_Fathur Rasyid Irimawa
```

#### Menentukan nilai x dan y

```python
X = df[['Per_Capita', 'GDP_Percentage']] 
y = df['External_Debt']

#202431036_Fathur Rasyid Irimawa
```

#### Bagi data latih dan data uji

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=0)
print(f"Jumlah data latih (train): {len(X_train)}")
print(f"Jumlah data uji (test): {len(X_test)}")

#202431036_Fathur Rasyid Irimawa
```

#### Model linearRegression

```python
model = LinearRegression()

#202431036_Fathur Rasyid Irimawa
```

#### Latih model dengan data latih

```python
model.fit(X_train, y_train)
print("Model berhasil dilatih!")

#202431036_Fathur Rasyid Irimawa
```

#### Prediksi data uji

```python
y_pred = model.predict(X_test)

#202431036_Fathur Rasyid Irimawa
```

#### Evaluasi model

```python
koefisien = model.coef_
intercept = model.intercept_
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"Koefisien (b1, b2) : {koefisien}")
print(f"Intercept (a)      : {intercept}")
print(f"MSE                : {mse}")
print(f"R2 Score           : {r2}")

#202431036_Fathur Rasyid Irimawa
```

#### Persamaan Regresi

```python
a = intercept
b1 = koefisien[0]
b2 = koefisien[1]

print("\n=== PERSAMAAN REGRESI ===")
print("Bentuk Persamaan: Y = a + b1*X1 + b2*X2")
print(f"Y = {a:.4f} + ({b1:.4f} * Per_Capita) + ({b2:.4f} * GDP_Percentage)")

#202431036_Fathur Rasyid Irimawa
```
